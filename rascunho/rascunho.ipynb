```python
import pandas as pd

df = pd.read_excel("dados.xlsx", sheet_name="Planejamento")
df.columns = df.columns.str.strip()

# Remove linhas que estão completamente vazias no trio de colunas
df = df.dropna(subset=['ID', 'Step Name' ,'Action', 'Expected Result'], how='all')

# Preenche NaN em ID com o valor anterior
df['ID'] = df['ID'].ffill()
df['ID'] = pd.to_numeric(df['ID'], errors='coerce').astype('Int64')

df.loc[:, ['ID', 'Step Name', 'Action', 'Expected Result']].head(25)
```
