# Rascunho dos processos de dados

```python
import pandas as pd

df = pd.read_excel("dados.xlsx", sheet_name="Planejamento")
```

```python
df.columns = df.columns.str.strip()
df.loc[:, ['ID','Import Test Case Name','Test Case Target', 'Card de Desenvolvimento', 'Sistema','Responsável','Link do Card de Desenvolvimento']].head(25)
```

```python
# Lista das colunas que quero
colunas_desejadas = ['ID','Import Test Case Name','Test Case Target', 'Card de Desenvolvimento', 'Sistema','Responsável','Link do Card de Desenvolvimento']

# Novo DataFrame só com essas colunas
df_tabela = df[colunas_desejadas].copy()

df_tabela = df_tabela.dropna(how='all') # Remove linhas onde todas as células são NaN
df_tabela["ID"] = df_tabela["ID"].astype("Int64")
df_tabela = df_tabela.apply(lambda col: col.apply(lambda x: x.strip() if isinstance(x, str) else x)) # Remove espaços em branco das células

df_tabela.head(50)
```
